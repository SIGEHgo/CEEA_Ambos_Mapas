```python
df_filtro = df_filtro.T.reset_index()
df_filtro.columns = df_filtro.iloc[0]
df_filtro = df_filtro.iloc[1:]
df_filtro.columns = df_filtro.columns.astype(str)
df_filtro = df_filtro.astype(str)

```

```python
def prevent_duplicates(cols):
    seen = {}
    new_cols = []
    for col in cols:
        if col in seen:
            seen[col] += 1
            new_cols.append(f"{col}.{seen[col]}")
        else:
            seen[col] = 0
            new_cols.append(col)
    return new_cols

df_filtro.columns = prevent_duplicates(df_filtro.columns)
```
